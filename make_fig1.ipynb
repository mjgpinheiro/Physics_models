```python
"""Fig. 1 from Tables III (20, 10, 5 dB) and IV (0 dB)."""
import numpy as np, matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
T3 = np.array([1e4, 3e4, 1e5]); T4 = np.array([1e4, 3e4, 1e5, 3e5])
data = {  # SNR: (T, ratio, YW, plateau)
    20: (T3, [0.0152, 0.0090, 0.0049], [0.0176, 0.0130, 0.0107], 0.0099),
    10: (T3, [0.0170, 0.0104, 0.0056], [0.0932, 0.0916, 0.0909], 0.0909),
    5:  (T3, [0.0257, 0.0153, 0.0083], [0.2425, 0.2410, 0.2403], 0.2403),
    0:  (T4, [0.0616, 0.0349, 0.0188, 0.0113],
             [0.5020, 0.5021, 0.5006, 0.5003], 0.5000)}
col = {20: "C0", 10: "C1", 5: "C2", 0: "C3"}
fig, ax = plt.subplots(figsize=(3.4, 2.9))
for s, (T, r, y, p) in data.items():
    ax.loglog(T, r, "o-", color=col[s], ms=4, lw=1.2)
    ax.loglog(T, y, "s--", color=col[s], ms=4, lw=1.2, mfc="white")
    ax.axhline(p, color=col[s], ls=":", lw=0.8)
Tg = np.array([7e3, 4e5])
ax.loglog(Tg, 0.0616 * (Tg / 1e4) ** -0.5, color="0.5", lw=0.8)
ax.text(1.6e5, 0.021, r"$T^{-1/2}$", color="0.4", fontsize=8)
ax.set_xlim(7e3, 4e5); ax.set_ylim(1.5e-3, 8)
ax.set_xlabel(r"record length $T$")
ax.set_ylabel(r"$\|\hat A-A\|_F/\|A\|_F$")
h1 = [Line2D([], [], color="k", marker="o", ms=4, label="ratio"),
      Line2D([], [], color="k", marker="s", ls="--", ms=4, mfc="white",
             label="Yule–Walker"),
      Line2D([], [], color="k", ls=":", label=r"plateau $\nu/(1+\nu)$")]
h2 = [Line2D([], [], color=col[s], lw=3, label=f"{s} dB") for s in data]
l1 = ax.legend(handles=h1, fontsize=6.5, loc="upper center", ncol=3, columnspacing=0.8, handlelength=1.6,
               bbox_to_anchor=(0.5, 1.0), frameon=False)
ax.add_artist(l1)
ax.legend(handles=h2, fontsize=6.5, loc="lower left", ncol=4,
          frameon=False, title="SNR", title_fontsize=6.5,
          columnspacing=0.8, handlelength=1.2)
ax.tick_params(labelsize=7); ax.xaxis.label.set_size(8); ax.yaxis.label.set_size(8)
fig.tight_layout(pad=0.3)
fig.savefig("fig1_errA_vs_T.pdf")

```
